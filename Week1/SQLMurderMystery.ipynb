select name from person as p
join drivers_license as d on p.license_id = d.id
join facebook_event_checkin as fec on p.id = fec.person_id
join income on p.ssn = income.ssn
where d.height between 65 and 67
and d.hair_color = 'red'
and d.car_model = 'Model S'
and fec.event_name = 'SQL Symphony Concert'
and fec.date between '20171201' and '20171231'
group by p.name, p.id having count (fec.event_name) = 3
order by income.annual_income desc;
